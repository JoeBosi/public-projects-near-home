# Primi passi con Python: i progetti pubblici di Arezzo

Dati: [OpenCoesione](https://opencoesione.gov.it/) (licenza CC BY 4.0). Premi il pulsante Play su ogni cella, dall'alto in basso. Non serve installare nulla: Google Colab esegue Python nel browser.

```python
import json, urllib.request

url = "https://opencoesione.gov.it/it/api/progetti/?format=json&territorio=arezzo-comune&page_size=2"
# Lo User-Agent dice al servizio chi sta chiedendo: un nome descrittivo è buona prassi
richiesta = urllib.request.Request(url, headers={"User-Agent": "progetti-didattico/1.0 (https://opendata.giuseppebosi.com)"})
with urllib.request.urlopen(richiesta) as risposta:
    dati = json.load(risposta)
```

La cella qui sopra chiede alla API due progetti del Comune di Arezzo. Quella che segue stampa il totale e, per ogni progetto, il CUP (codice del progetto), lo stato, il finanziato e il pagato.

```python
print("Progetti trovati:", dati["count"])
for p in dati["results"]:
    print(p["cup"], p["oc_stato_progetto"], p["oc_finanz_tot_pub_netto"], p["tot_pagamenti"])
```
