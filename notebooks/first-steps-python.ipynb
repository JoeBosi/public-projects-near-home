# First steps with Python: public projects of Arezzo

Data: [OpenCoesione](https://opencoesione.gov.it/) (licence CC BY 4.0). Press the play button on each cell, from top to bottom. Nothing to install: Google Colab runs Python in your browser.

```python
import json, urllib.request

url = "https://opencoesione.gov.it/it/api/progetti/?format=json&territorio=arezzo-comune&page_size=2"
# The User-Agent tells the service who is asking: a descriptive name is good practice
request = urllib.request.Request(url, headers={"User-Agent": "projects-tutorial/1.0 (https://opendata.giuseppebosi.com)"})
with urllib.request.urlopen(request) as response:
    data = json.load(response)
```

The cell above asks the API for two projects of the Municipality of Arezzo. The next one prints the total and, for each project, the CUP (project code), the status, the funded amount and the paid amount.

```python
print("Projects found:", data["count"])
for p in data["results"]:
    print(p["cup"], p["oc_stato_progetto"], p["oc_finanz_tot_pub_netto"], p["tot_pagamenti"])
```
